```python
employee_df["skills_clean"] = employee_df["skills"].str.replace(";", " ")
team_df["skills_clean"] = team_df["required_skills"].str.replace(";", " ")
```

```python
employee_df["profile_text"] = (
    employee_df["job_role"].astype(str) + " " +
    employee_df["skills_clean"].astype(str) + " " +
    employee_df["domain"].astype(str)
)
```

```python
employee_df["profile_text"][700]
```

```python
team_df["team_profile_text"] = (
    team_df["job_role"].astype(str) + " " +
    team_df["skills_clean"].astype(str) + " " +
    team_df["domain"].astype(str)
)
```

```python
team_df["team_profile_text"][0]
```

```python
team_df.tail()
```

```python
employee_df.tail()
```

```python
model = SentenceTransformer("all-mpnet-base-v2")
```

```python
employee_embeddings = model.encode(
    employee_df["profile_text"],
    batch_size=32,
    show_progress_bar=True,
    convert_to_tensor=True,
    normalize_embeddings=True
)
```

```python
employee_embeddings.shape
```

```python
team_embeddings = model.encode(
    team_df["team_profile_text"],
    batch_size=32,
    show_progress_bar=True,
    convert_to_tensor=True,
    normalize_embeddings=True
)
```

```python
team_embeddings.shape
```

```python
print(team_df["team_profile_text"].iloc[0])
print(team_embeddings[0][:10])
```

```python
print(employee_df["profile_text"].iloc[0])
print(employee_embeddings[0])
```

```python
import numpy as np
```

```python
np.save("team_embeddings.npy", team_embeddings.cpu())
np.save("employee_embeddings.npy", employee_embeddings.cpu())
```

```python
similarity_matrix = np.dot(employee_embeddings, team_embeddings.T)
```

```python
similarity_matrix.shape
```

```python
print(similarity_matrix.shape)
print(similarity_matrix[0])

```

```python
np.save("similarity_matrix.npy", similarity_matrix)
```

```python
similarity_df = pd.DataFrame(
    similarity_matrix,
    columns=team_df["team_id"] + "_" + team_df["job_role"]
)
```

```python
print(similarity_matrix.shape)
print(len(employee_df))
print(len(team_df))
```

```python
print(employee_df.shape)
print(team_df.shape)

print(team_df["team_id"].nunique())
```

```python
#For every employee, find the highest similarity score.
```

```python
best_match_index = similarity_matrix.argmax(axis=1)

best_match_scores = similarity_matrix.max(axis=1)

best_roles = team_df.iloc[best_match_index]["job_role"].values
best_teams = team_df.iloc[best_match_index]["team_id"].values

employee_match_df = pd.DataFrame({
    "employee_id": employee_df["employee_id"],
    "best_team": best_teams,
    "best_role": best_roles,
    "similarity_score": best_match_scores
})

print(employee_match_df.head())
```

```python
#which employees best fit each role.
```

```python
best_employee_index = similarity_matrix.argmax(axis=0)

best_employee_scores = similarity_matrix.max(axis=0)

role_match_df = pd.DataFrame({
    "team_id": team_df["team_id"],
    "role": team_df["job_role"],
    "best_employee": employee_df.iloc[best_employee_index]["employee_id"].values,
    "similarity_score": best_employee_scores
})

print(role_match_df.head())
```
